```python
import os, sys
from pathlib import Path
if Path.cwd().name == "notebooks":
    os.chdir(Path.cwd().parent)
sys.path.insert(0, str(Path.cwd()))

print(f"Raíz: {Path.cwd()}")

# Borrar informe.qmd accidental
accidental = Path("reports/informe.qmd")
if accidental.exists() and accidental.stat().st_size == 0:
    accidental.unlink()
    print(f"🗑️  Borrado: {accidental}")
else:
    print("ℹ️  No había informe.qmd vacío")

# Crear carpetas
Path("informe/figuras").mkdir(parents=True, exist_ok=True)
Path("data/processed").mkdir(parents=True, exist_ok=True)
print("✅ Carpetas creadas: informe/, informe/figuras/, data/processed/")
```
