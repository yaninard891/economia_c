```python
from pathlib import Path
import pandas as pd

RAIZ = Path.cwd()
if RAIZ.name == "notebooks":
    RAIZ = RAIZ.parent

RUTA = RAIZ / "data" / "raw" / "provinciales_serie_empresas3_2.xlsx"
df = pd.read_excel(RUTA, sheet_name="La Rioja", header=None)

# Fila de años = 6, años arrancan en columna 2
anios = df.iloc[6].tolist()
idx_1996 = next(j for j, v in enumerate(anios) if str(v) == "1996")
anios_validos = [a for a in anios[idx_1996:] if str(a) != 'nan']

# Sumar todas las filas que tengan un código CIIU numérico
totales = {a: 0 for a in anios_validos}
for i in range(7, len(df)):
    try:
        cod = int(df.iloc[i, 0])
    except (ValueError, TypeError):
        continue
    valores = df.iloc[i, idx_1996:].tolist()
    for anio, v in zip(anios_validos, valores):
        try:
            totales[anio] += float(v)
        except (ValueError, TypeError):
            pass  # ignora s.d., nan, texto

# Mostrar
print("=== TOTAL PROVINCIAL LA RIOJA ===")
for a in anios_validos:
    print(f"  {a}: {int(totales[a]):,}")

# Guardar CSV
out = pd.DataFrame({
    "anio": anios_validos,
    "total_provincial": [int(totales[a]) for a in anios_validos],
})
destino = RAIZ / "data" / "raw" / "total_provincial_rioja.csv"
out.to_csv(destino, index=False)
print(f"\nGuardado en: {destino}")
```
